```python
def stretch(X: np.array, a, b):
    A = np.array([[a, 0], [0, b]])

    if X.shape[0] != A.shape[1]:
        raise ValueError("Can't multiplie matrixes")

    x_copy = X.copy()

    return (A @ x_copy, A)
```

```python
X = np.array([[1, 2],
              [3, 4]])

X_streched = stretch(X, 2, 4)[0]

print(X_streched)
```

```python
def shear(X, a, b):
    A = np.array([[1, b], [a, 1]])

    if X.shape[0] != A.shape[1]:
        raise ValueError("Can't multiplie matrixes")

    x_copy = X.copy()

    return (A @ x_copy, A)
```

```python
X = np.array([[1, 2],
              [3, 4]])

X_sheared= shear(X, 2, 4)[0]

print(X_sheared)
```

```python
batman_sheared = shear(batman, 0.5, 0.5)

print(batman_sheared[1])
visalizing_2d(batman, batman_sheared[0], batman_sheared[1], 'Sheared')
```

```python
def reflection(X, a, b):
    A = 1/(a**2 + b**2) *  np.array([[a**2 - b**2, 2*a*b], [2*a*b, b**2 - a**2]])

    if X.shape[0] != A.shape[1]:
        raise ValueError("Can't multiplie matrixes")

    x_copy = X.copy()

    return (A @ x_copy, A)
```

```python
X = np.array([[1, 2],
              [3, 4]])

X_reflected = reflection(X, 2, 4)[0]

print(X_reflected)
```

```python
lynx_reflected = reflection(lynx, 0, 1)

print(lynx_reflected[1])
visalizing_2d(lynx, lynx_reflected[0], lynx_reflected[1], 'Reflected')
```

```python
def rotation(X, fi):
    A = np.array([[np.cos(fi), -1 * np.sin(fi)], [np.sin(fi), np.cos(fi)]])

    if X.shape[0] != A.shape[1]:
        raise ValueError("Can't multiplie matrixes")

    x_copy = X.copy()

    return (A @ x_copy, A) 
```
